```python
import yaml
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy import text


pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)

# Leer credenciales del arhivo YAML
with open("credenciales.yaml", "r") as file:
     config = yaml.safe_load(file)

db = config["mysql"]

# Crear conexión
# Su función es crear un Engine, que es el objeto que SQLAlchemy utiliza para comunicarse con la base de datos.
engine = create_engine(                                     
    f"mysql+pymysql://{db['user']}:{db['password']}@"
    f"{db['host']}:{db['port']}/{db['database']}"
)
```

```python
# Leer CSV
df = pd.read_csv("C:/Users/nickc/Downloads/Archivos de trabajo - SQL para análisis de datos/Archivos Alumnos/Cap.4_Base de datos/Cap. 4 Base de datos/legisladoresusa.csv")

# Tranformar datos si es necesario, en este caso conertir columna a tipo date
# Convertir fechas
df["InicioPeriodo"] = pd.to_datetime(df["InicioPeriodo"])
df["FinPeriodo"] = pd.to_datetime(df["FinPeriodo"])

# 4. Cargar datos a la tabla legisladoresusa en el esquema cohort
df.to_sql(
    name="legisladoresusa",
    con=engine,
    schema="cohort",
    if_exists="append",
    index=False
)
```

```python
# En caso de ser necesario podemos vaciar la tabla
with engine.begin() as conn:
    conn.execute(text("TRUNCATE TABLE cohort.legisladoresusa"))
```
