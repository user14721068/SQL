# **¿Qué son las conexiones?**

Una conexión en MySQL es un canal de comunicación interactivo y autenticado que se establece entre un cliente (como una aplicación, un script, la línea de comandos o una herramienta visual como MySQL Workbench) y el servidor de base de datos MySQL.

A través de esta conexión, el cliente puede enviar consultas (queries) en lenguaje SQL y recibir los resultados de las tablas o bases de datos alojadas en el servidor.

Elementos principales de una conexión

Para que una conexión se pueda establecer con éxito, generalmente se requieren los siguientes parámetros:

 - Host / Dirección IP: La ubicación del servidor (por ejemplo, localhost o 127.0.0.1 si está en la misma máquina, o una IP remota).
 - Puerto: El canal de comunicación por defecto que utiliza MySQL, el cual suele ser el 3306.
 - Nombre de usuario: La cuenta de MySQL con permisos asignados (por ejemplo, root o un usuario específico).
 - Contraseña: Las credenciales de seguridad asociadas a dicho usuario.
 - Base de datos (Opcional): El esquema o conjunto de tablas específico al que se desea apuntar de entrada tras autenticarse.


¿Cómo funciona en la práctica?

 1. Solicitud: El cliente (una web en PHP, Python, Java o un programa de escritorio) utiliza un controlador o conector (Connector/Python, JDBC, PDO, etc.) para llamar a una función de conexión.
 2. Autenticación: El servidor MySQL verifica las credenciales y los privilegios del usuario.
 3. Interacción: Una vez aceptada, se abre el canal para ejecutar operaciones (SELECT, INSERT, UPDATE, DELETE).
 4. Cierre: Al terminar las operaciones, se debe cerrar la conexión (close()) para liberar los recursos del servidor

# **Crear conexión en MySQL Workbench**

Para conectar MySQL Workbench a una base de datos local (o remota), sigue estos pasos desde la interfaz del programa:

Conexión local (en tu propia computadora)

 1. Abre MySQL Workbench en tu computadora.
 2. Haz clic en el icono de más (+) que se encuentra junto a la sección MySQL Connections en la pantalla de inicio.
 3. Rellena los datos básicos en la ventana que aparece (puedes guiarte por el Tutorial de inicio rápido de MySQL):
   - Connection Name: Asigna un nombre a la conexión (por ejemplo, MiConexionLocal).
   - Connection Method: Déjalo en Standard TCP/IP.
   - Hostname: 127.0.0.1 o localhost.
   - Port: 3306 (puerto predeterminado)
   - Username: root (o el usuario que hayas creado).
 4. Haz clic en el botón Test Connection para comprobar que los datos sean correctos.
 5. Te pedirá la contraseña del usuario (la que definiste al instalar MySQL Server); puedes guardarla en el almacén de claves (Vault) para no escribirla cada vez.
 6. Haz clic en OK y luego en la nueva tarjeta de conexión creada para entrar al panel de administración.

# **Importar datos**

Para importar una base de datos desde un archivo Self-Contained File (.sql) en Windows, el método más directo y visual es mediante MySQL Workbench, aunque también puedes hacerlo desde la consola de comandos si el archivo es muy pesado.

Método 1. Usando MySQL Workbench (Interfaz gráfica)

Este es el método estándar diseñado específicamente para archivos autocontenidos.
 - Paso 1. Abre tu conexión. Inicia MySQL Workbench en Windows y haz clic sobre tu instancia local de MySQL (ingresa tu contraseña si te la solicita).
 - Paso 2. Accede a la herramienta de importación. En el menú superior, haz clic en Server y luego selecciona Data Import (también lo encuentras en el panel izquierdo bajo la pestaña Administration).
 - Paso 3. Selecciona el archivo. Marca la casilla Import from Self-Contained File. Haz clic en el botón de los tres puntos (...) a la derecha y busca tu archivo .sql en tu computadora.
 - Paso 4. Elige la base de datos de destino. En el apartado Default Target Schema, selecciona el esquema donde se cargarán los datos. Si el archivo no crea la base de datos por sí solo, haz clic en el botón New... para crear una vacía en ese momento.
 - Paso 5. Ejecuta la importación. Ve a la pestaña Import Progress (en la parte inferior o superior según tu versión) y haz clic en el botón Start Import.
 - Paso 6. Verifica los datos. Una vez que aparezca el mensaje de finalización, ve al panel izquierdo de Schemas, haz clic en el ícono de actualizar (las dos flechas circulares) y confirma que tus tablas estén ahí.


Método 2. Usando la consola de Windows (Para archivos grandes)

Si tu archivo .sql pesa cientos de megabytes o gigabytes, MySQL Workbench puede congelarse. En ese caso, usa el Símbolo del sistema (CMD).
 - Paso 1. Ubica la carpeta bin de MySQL. Abre el explorador de archivos y ve a la ruta de instalación de tu servidor (por ejemplo, C:\Program Files\MySQL\MySQL Server 8.0\bin o C:\xampp\mysql\bin).
 - Paso 2. Abre la consola en esa ruta. Escribe cmd en la barra de direcciones de esa carpeta y presiona Enter.
 - Paso 3. Crea la base de datos (si no existe). Ejecuta mysql -u root -p, ingresa tu contraseña y escribe CREATE DATABASE nombre_base_datos; seguido de EXIT;.
 - Paso 4. Importa el archivo. En la misma ventana de comandos, ejecuta la siguiente instrucción reemplazando los datos por los tuyos:
mysql -u root -p nombre_base_datos < "C:\ruta\de\tu\archivo.sql"
 - Paso 5. Espera a que termine. Escribe tu contraseña y presiona Enter. El cursor parpadeará sin mostrar texto hasta que toda la carga finalice y vuelva a aparecer la ruta de la carpeta.

# **Intro a Tipos de Datos** 

En MySQL, un tipo de datos es una regla de definición que se le asigna a cada columna de una tabla para indicar qué clase de valores puede almacenar (números, texto, fechas, etc.) y cuánto espacio de memoria ocuparán.
Define cómo el sistema gestor de bases de datos valida, almacena, ordena, compara y realiza operaciones matemáticas con la información.

Es importante entender la diferencia entre tipo de dato y formato del dato: el tipo de dato define la naturaleza y el almacenamiento físico de la información a nivel de motor de base de datos, mientras que el formato de dato es solo la manera visual o de representación en que se muestra al usuario.

El formato de dato es la apariencia, el orden o la máscara con la que se presentan esos datos cuando se leen (o cómo se interpretan al ingresarlos). Es superficial o de presentación; la base de datos almacena el valor de forma estandarizada independientemente de cómo lo imprimas. Ejemplos en MySQL:
 - Para una fecha almacenada como DATE (que internamente se guarda como YYYY-MM-DD), puedes usar funciones como DATE_FORMAT() para mostrarla visualmente como DD-MM-YYYY o 15 de diciembre de 2026.
 - Formatear un número con separadores de miles o decimales específicos al momento de hacer un SELECT.
 - Impacto: No cambia cómo se guarda el dato en el disco duro, pero consume tiempo de procesamiento extra si tienes que formatear millones de registros al vuelo en cada consulta.

 ¿Qué tipos de datos tenemos?

 - String / Cadenas de texto: Se utilizan para almacenar texto, palabras, frases, símbolos o incluso datos binarios. Ejemplos son CHAR(n), VARCHAR(n), TEXT, BINARY, VARBINARY, BLOB, ENUM, SET.
 - Numéricos
 - Tiempo / Fecha
 - BLOB / Binarios
 - Espaciales



```python
import yaml
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy import text


pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)

# Leer credenciales del arhivo YAML
with open("credenciales.yaml", "r") as file:
     config = yaml.safe_load(file)

db = config["mysql"]

# Crear conexión
# Su función es crear un Engine, que es el objeto que SQLAlchemy utiliza para comunicarse con la base de datos.
engine = create_engine(                                     
    f"mysql+pymysql://{db['user']}:{db['password']}@"
    f"{db['host']}:{db['port']}/{db['database']}"
)
```

```python
# Leer CSV
df = pd.read_csv("C:/Users/nickc/Downloads/Archivos de trabajo - SQL para análisis de datos/Archivos Alumnos/Cap.4_Base de datos/Cap. 4 Base de datos/TablaFechas.csv")
df.head(10)
```

```python
# Cargar datos a la tabla legisladoresusa en el esquema cohort
df.to_sql(
    name="tablafechas",
    con=engine,
    schema="cohort",
    if_exists="append",
    index=False
)
```

```python
# En caso de ser necesario podemos vaciar la tabla
with engine.begin() as conn:
    conn.execute(text("TRUNCATE TABLE cohort.tablafechas"))
```

```python
# También podemos eliminar la tabla

with engine.begin() as conn:
    conn.execute(text("DROP TABLE cohort.tablafechas;"))
```
